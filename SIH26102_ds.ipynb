**Feature Engineering**

```python
# Expenditure Ratio: Calculate how much of the sanctioned amount has been spent
final["expenditure_ratio"] = final["Fund Disbursed Amount ( ₹ )"] / final["Sanction Amount ( ₹ )"]
```

```python
# View the expenditure ratio
final[["work_id", "Sanction Amount ( ₹ )", "Fund Disbursed Amount ( ₹ )", "expenditure_ratio"]].head()
```

```python
# Remaining Amount: the amount left after expenditure
final["remaining_amount"] = final["Sanction Amount ( ₹ )"] - final["Fund Disbursed Amount ( ₹ )"]
```

```python
# Check the remaining amount
final[["work_id", "Sanction Amount ( ₹ )", "Fund Disbursed Amount ( ₹ )", "remaining_amount"]].head()
```

```python
# Convert sanction and completion dates to date format
final["Sanction Date"] = pd.to_datetime(final["Sanction Date"])
final["Completion Date"] = pd.to_datetime(final["Completion Date"])
```

```python
# Convert dates using the known date format
final["Sanction Date"] = pd.to_datetime(final["Sanction Date"], format="%d-%b-%y", errors="coerce")
```

```python
# Convert completion dates
final["Completion Date"] = pd.to_datetime(final["Completion Date"], format="%d-%b-%Y", errors="coerce")
```

```python
# Execution time: how long a work takes from sanction to completion.
# Completion time = Completion Date − Sanction Date
```

```python
# Calculate days taken to complete each work
final["completion_days"] = (final["Completion Date"] - final["Sanction Date"]).dt.days
```

```python
# Check completion duration
final[["work_id", "Sanction Date", "Completion Date", "completion_days"]].head()
```

```python
# sanction amount vs expenditure
# Identify works where expenditure exceeds sanctioned amount
final["over_expenditure"] = final["expenditure_ratio"] > 1
```

```python
# Flag only works with available expenditure data
final["over_expenditure"] = final["expenditure_ratio"].where(
    final["expenditure_ratio"].isna(),
    final["expenditure_ratio"] > 1
)
```

```python
# Count each over-expenditure status
final["over_expenditure"].value_counts(dropna=False)
```

```python
# Convert recommended dates to proper date format
final["Recommended date"] = pd.to_datetime(final["Recommended date"], format="%d-%b-%y", errors="coerce")
```

```python
# completion delay: a project without a completion date may still be running.
# Calculate days between recommendation and sanction
final["sanction_delay_days"] = (final["Sanction Date"] - final["Recommended date"]).dt.days
```

```python
# Check the calculated delay
final[["work_id", "Recommended date", "Sanction Date", "sanction_delay_days"]].head()
```

```python
# Understanding the overall delay distribution
# Check basic statistics of sanction delays
final["sanction_delay_days"].describe()
```

```python
# Flag works with unusually high sanction delay
final["long_sanction_delay"] = final["sanction_delay_days"] > 140
```

```python
# Count long and normal sanction delays
final["long_sanction_delay"].value_counts()
```

```python
# Check how long completed works usually take
final["completion_days"].describe()
```

```python
# Count payment transactions for each work
payment_count = expenditure.groupby("work_id").size()
```

```python
# Count different vendors for each work
vendor_count = expenditure.groupby("work_id")["Vendor Name"].nunique()
```

```python
# Add vendor count to the master dataset
final = final.merge(vendor_count.rename("vendor_count"), on="work_id", how="left")
```

```python
# Calculate the average payment for each work
avg_payment = expenditure.groupby("work_id")["Fund Disbursed Amount ( ₹ )"].mean()
```

```python
# Add average payment to the master dataset
final = final.merge(avg_payment.rename("avg_payment"), on="work_id", how="left")
```

```python
# Find the largest payment made for each work
max_payment = expenditure.groupby("work_id")["Fund Disbursed Amount ( ₹ )"].max()
```

```python
# Create the dataset that will be given to the ML model
X = final[ml_features].copy()
```

Standardization

```python
# Standardize all ML features to a comparable scale
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Check the shape of the scaled data
X_scaled.shape
```

**Isolation Forest anomaly detection model**

```python
# Create the Isolation Forest anomaly detection model
from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=200,
    contamination=0.05,
    random_state=42
)
```

```python
# Train Isolation Forest on the prepared MPLADS data
model.fit(X_scaled)
```

Generating Anomaly Predictions

```python
# Generate anomaly predictions for every work
final["anomaly_prediction"] = model.predict(X_scaled)
```

```python
# Count normal and anomalous works
final["anomaly_prediction"].value_counts()
```

Anomaly Score

```python
# Calculate how unusual each work is
final["anomaly_score"] = model.decision_function(X_scaled)
```

```python
# View the most unusual works
final.sort_values("anomaly_score").head()
```

Creating an overall risk score

```python
# Calculate an explainable risk score from the warning indicators
risk_flags = [
    "high_cost_flag",
    "long_duration_flag",
    "long_sanction_delay",
    "over_expenditure",
    "high_expenditure_flag",
    "late_first_payment_flag"
]

final["risk_score"] = final[risk_flags].fillna(False).astype(int).sum(axis=1) / len(risk_flags) * 100
```

```python
# View the distribution of the explainable risk score
final["risk_score"].value_counts().sort_index()
```

**Human-readable risk categories** based on the number of triggered indicators: 

0–1 indicators → Low Risk;
2 indicators → Medium Risk;
3–4 indicators → High Risk;
5–6 indicators → Critical Risk;

0–16.67 → Low;
33.33 → Medium;
50–66.67 → High;
83.33–100 → Critical;

```python
# Correct risk categories using the actual score ranges
def assign_risk_category(score):
    if score <= 16.67:
        return "Low Risk"
    elif score <= 33.34:
        return "Medium Risk"
    elif score <= 66.67:
        return "High Risk"
    else:
        return "Critical Risk"

final["final_risk_category"] = final["risk_score"].apply(assign_risk_category)
```

```python
# Count works in each final risk category
final["final_risk_category"].value_counts()
```

```python
# Compare Isolation Forest results with our explainable risk categories
pd.crosstab(
    final["final_risk_category"],
    final["anomaly_prediction"]
)
```

Calculating AI/rule agreement

```python
# Check how many AI-anomalous works also have high/critical rule-based risk
ai_anomalous = final["anomaly_prediction"] == -1
rule_high_risk = final["risk_score"] >= 50

agreement_count = (ai_anomalous & rule_high_risk).sum()

print("AI anomalies:", ai_anomalous.sum())
print("High/Critical rule-based risks:", rule_high_risk.sum())
print("Both AI + rules flagged:", agreement_count)
```

Creating the final priority alert flag

```python
# Create a final priority alert when both methods identify risk
final["priority_alert"] = (
    (final["anomaly_prediction"] == -1) &
    (final["risk_score"] >= 50)
)
```

```python
# Count priority alerts
final["priority_alert"].value_counts()
```

Creating a final alert type

```python
# Create a final alert type by combining AI and rule-based results
def assign_alert_type(row):
    if row["priority_alert"]:
        return "Priority Alert"
    elif row["anomaly_prediction"] == -1:
        return "AI Anomaly"
    elif row["risk_score"] >= 50:
        return "Rule-Based High Risk"
    else:
        return "Normal"

final["alert_type"] = final.apply(assign_alert_type, axis=1)
```
